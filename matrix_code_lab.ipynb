## 1. Shape of a matrix
**Try first:** predict the shape and total number of elements.

```python
import numpy as np
import pandas as pd

A = np.array([[1, 2, 3], [4, 5, 6]])
print(A)
print('Shape:', A.shape)
print('Elements:', A.size)
```

## 2. Common matrix types
Complete the examples and explain the pattern in each matrix.

```python
row_matrix = np.array([[1, 2, 3]])
column_matrix = np.array([[1], [2], [3]])
zero_matrix = np.zeros((2, 3), dtype=int)
identity_matrix = np.eye(3, dtype=int)
diagonal_matrix = np.diag([2, 5, 8])

print('Row:', row_matrix, row_matrix.shape)
print('Column:\n', column_matrix, column_matrix.shape)
print('Zeros:\n', zero_matrix)
print('Identity:\n', identity_matrix)
print('Diagonal:\n', diagonal_matrix)
```

```python
ones_matrix = np.ones((2,3))
print(ones_matrix)
```

## 3. Transpose
Predict how rows and columns change before running the next cell.

```python
print('A shape:', A.shape)
print('A transpose:\n', A.T)
print('A.T shape:', A.T.shape)
```

## 4. ML-style dataset with pandas
Rows are students and columns are features in this example.

```python
df = pd.DataFrame({
    'attendance': [80, 65, 95],
    'study_hours': [6, 3, 8],
    'assignment_score': [90, 70, 96]
}, index=['Student A', 'Student B', 'Student C'])
display(df)
X = df.to_numpy()
print('Shape:', X.shape)
print('Feature means:', X.mean(axis=0))
print('Transpose:\n', X.T)
print('Transpose shape:', X.T.shape)
```
